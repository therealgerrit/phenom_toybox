```python
import numpy as np
from matplotlib import pyplot as plt
from phenom.source import sase_pulse
from phenom.spectrum import linear_SASE_spectrum
```

```python
# Example usage with plotting enabled
t = np.linspace(-25e-15, 25e-15, 1500)
spectrum = linear_SASE_spectrum(t, pulse_duration=5e-15, photon_energy=9500, bandwidth=1e-12, plot=True)
```

```python
## define the spatial/temporal mesh
x = y = np.linspace(-250e-06, 250e-06, 500)
t = np.linspace(-25e-15, 25e-15, 250)

sase = linear_SASE_spectrum(t, 5e-03 , 5e-15, 9200)

## define the pulse
pulse = sase_pulse(x = x,
                   y = y,
                   t = t,
                   photon_energy = 9200.,
                   pulse_energy = 1e-04,
                   pulse_duration = 5e-15,
                   bandwidth = 1e-12,
                   sigma = 50e-06,
                   div = 2.5e-03,
                   x0 = 0.0,
                   y0 = 0.0,
                   t0 = 0.0,
                   theta_x = 0.0,
                   theta_y = 0.0
                  )
```

```python
fig, [ax1, ax2] = plt.subplots(1,2, figsize = (9,3))

### spatial intensity
im = ax1.imshow((abs(pulse)**2).sum(axis = 2),
                extent = [x.min() * 1e06,
                          x.max() * 1e06,
                          y.min() * 1e06,
                          y.max() * 1e06],
                cmap = 'bone'
                )

plt.colorbar(im, label = "Intensity (W/mm$^2$)", ax = ax1)

ax1.set_xlabel('x ($\\mu$m)')
ax1.set_ylabel('y ($\\mu$m)')


ax2.plot(t*1e15, (abs(pulse)**2).sum(axis = (0,1)))
ax2.set_xlabel("Time (fs)")
```
